## Numpy array and basics

```python
import numpy as np
```

### creating array form list 

```python
arr_1d = np.array([1,2,3,4,5])
print("1d array: ", arr_1d)
```

### list vs numpy array

```python
py_list = [1,2,3,4]
print("python list multiplication", py_list*2)

np_array =  np.array([1,2,3,4])
print("python array multiplication :", np_array*2)

import time
start = time.time()
py_list = [i*2 for i in range(1000)]
print("\n list opertaion time :", time.time()-start)

start = time.time()
np_array = np.arange(1000)*2
print("\n array opertaion time :", time.time()-start)
```

### creating array from scratch

```python
zeros = np.zeros((3,4))
print("zeros array: \n",zeros)

ones = np.ones((3,4))
print("ones array: \n",ones)

full = np.full((2,2),7)
print("full array: \n",full)

random = np.random.random((2,3))
print("random value array: \n",random)

sequence = np.arange(0,10,2)
print("sequence value array: \n",sequence)
```

#### Vector , matrix, Tensor

```python
vector = np.array([1,2,3])
print("array : \n", vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("matrix : \n", matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor : \n", tensor)
```

### Array Properties 

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print("shape",arr.shape)
print("dimension",arr.ndim)
print("size",arr.size)
print("DType",arr.dtype)
```

Array reshaping

```python
arr = np.arange(12)
print("Original array \n ",arr)

reshape = arr.reshape((3,4))
print("reshaped array \n",reshape)

flatten = reshape.flatten()
print("flattened array \n", flatten)

# Ravel : return view instead of copy
ravel = reshape.ravel()
print("ravel array \n", ravel)

# Transpose 
transpose = reshape.T
print("Transpose array \n", transpose)
```
