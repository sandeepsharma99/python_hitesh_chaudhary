```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# DATASET : [restaurent_id,2021,2022,2023,2024]
sales_data = np.array([
    [1,150000,180000,220000,350000], #paradise Biryani
    [2,120000,140000,160000,190000], # Beijing Bites
    [3,200000,230000,260000,300000], # Pizza Hub
    [4,180000,210000,240000,270000], # Burger Point
    [5,160000,185000,205000,230000], # Chai Point
])

print("===== Zomato sales Analysis =====" )
print("sales data shape :\n", sales_data.shape)
print("first 3 restaturent :\n",sales_data[:,1:])
```

```python
# Total slaes per year
print(np.sum(sales_data,axis=0)) # 0 -> top to bottom , 1 -> left to right
print(np.sum(sales_data[:,1:], axis = 0))

# Minimum sales per year
minimum_sale = np.min(sales_data[:,1:],axis=1)
print(minimum_sale)

#Maximum sales per year
max_sale = np.max(sales_data[:,1:], axis=1)
print(max_sale)

# Average sale per year
avg_sale = np.mean(sales_data[:,1:],axis=1)
print(avg_sale)
```

```python
# cumulative sales
cumulativ_sale = np.cumsum(sales_data[:,1:], axis=1)
print(cumulativ_sale)
plt.figure(figsize=(7,3))
plt.plot(np.mean(cumulativ_sale,axis=0))
plt.title("average cumulative sale across all restaurent")
plt.xlabel("year")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
